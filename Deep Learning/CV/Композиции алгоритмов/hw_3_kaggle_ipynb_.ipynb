```python
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, RobustScaler, LabelEncoder, OneHotEncoder, MinMaxScaler
from sklearn.pipeline import make_pipeline
from sklearn.compose import ColumnTransformer
import catboost
```

```python
# категориальные и числовые столбцы
cat_columns = ['HasInternetService', 'HasContractPhone', 'PaymentMethod']
num_columns = [i for i in data.columns if i not in cat_columns]
```

```python
# кодирование
dummy_features = pd.get_dummies(data[cat_columns])
data_dum = pd.concat([data[num_columns], dummy_features], axis=1)
data_dum.shape
```

```python
# разделение
X, y = data_dum.drop('Churn', axis=1), data['Churn']

X_train, X_val, y_train, y_val = train_test_split(X, y, train_size=0.8,
                                                       random_state=19)
```

```python
# обучение
pipeline = make_pipeline(
    StandardScaler(), # нормирование
    LogisticRegression(random_state=19)
)
# baseline
lr_base = pipeline.fit(X_train, y_train)
print(f'ROC-AUC на тестовой выборке baseline: {roc_auc_score(y_val, lr_base.predict_proba(X_val)[:, 1])}')

# подбор гиперпараметров
parameters = {'logisticregression__C': [1000, 100, 10, 1, 0.1, 0.01, 0.001],
              'logisticregression__max_iter': [10000, 5000, 1000, 500],
              'logisticregression__solver': ['lbfgs', 'liblinear', 'newton-cg', 'newton-cholesky', 'sag', 'saga']
              }
lr = GridSearchCV(estimator=pipeline,
                            param_grid=parameters,
                            cv=5,
                            scoring='roc_auc',
                            refit=True,
                            n_jobs=-1)
lr.fit(X_train, y_train)

print(f'ROC-AUC на тестовой выборке: {roc_auc_score(y_val, lr.predict_proba(X_val)[:, 1])}')
```

Мы будем использовать catboost, поэтому нам не надо кодировать категориальные признаки. Catboost сделает это сам (в .fit() надо передать cat_features=cat_cols). А численные признаки нормировать для моделей, основанных на деревьях не нужно.

1) Разделите выборку на train/valid. Протестируйте catboost cо стандартными параметрами.

2) Протестируйте разные занчения параметров количества деревьев и learning_rate'а и выберите лучшую по метрике ROC-AUC комбинацию.

(Дополнительно) Есть некоторые сложности с тем, чтобы использовать CatBoostClassifier вместе с GridSearchCV, поэтому мы не просим использовать кроссвалидацию. Но можете попробовать)

```python
X, y = data.drop('Churn', axis=1), data['Churn']
X_train, X_val, y_train, y_val = train_test_split(X, y, train_size=0.8,
                                                       random_state=19)

boosting_model = catboost.CatBoostClassifier(n_estimators=200,
                                             cat_features=cat_columns_test,
                                             verbose = False)

boosting_model.fit(X_train, y_train)

print(f'ROC-AUC на тестовой выборке: {roc_auc_score(y_val, boosting_model.predict_proba(X_val)[:, 1])}')
```

```python
# обучение
parameters = {'iterations': [200, 300, 400],
              'learning_rate': np.linspace(0.01, 0.5, 8),
              'depth': [4, 6, 8, 10]
              }
cb = GridSearchCV(estimator=catboost.CatBoostClassifier(
                                             cat_features=cat_columns_test,
                                             verbose = False,
                                             thread_count=1,
                                             random_state=42),
                            param_grid=parameters,
                            cv=5,
                            scoring='roc_auc',
                            refit=True,
                            n_jobs=-1)
cb.fit(X_train, y_train)

print(f'ROC-AUC на тестовой выборке: {roc_auc_score(y_val, cb.predict_proba(X_val)[:, 1])}')
```
