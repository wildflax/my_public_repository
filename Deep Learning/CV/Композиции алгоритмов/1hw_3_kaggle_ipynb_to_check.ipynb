В таких случаях нужно 1) не использовать accuracy как метрику 2) использовать методы борьбы с imbalanced dataset (обычно если датасет сильно несбалансирован, т.е. класса 1 в 20 раз меньше класса 0).

```python
# Ф-ция для построения круговых диаграмм
def generate_pies(data, column):
    n = data[column].nunique() # уникальных значений в столбце
    u_values = data[column].unique()
    n_cols = 4
    n_rows = (n + n_cols - 1) // n_cols

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 4 * n_rows))
    axes = axes.flatten()
# для каждого уникального значения строим диаграмму зависимости от таргета
    for i in range(n):
        data_pie = data[data[column] == u_values[i]][target_col].value_counts()
        axes[i].pie(data_pie.values, labels=data_pie.index, autopct='%1.1f%%')
        axes[i].set_title(f'{u_values[i]}')
# скрываем лишние оси
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)


    fig.legend(['нет', 'отток'], title=target_col, fontsize=7)
    plt.suptitle(column)
    plt.tight_layout()
    plt.show()
```

```python
# Построим круговые диаграммы по всем категориальным признакам, уникальных значений не больше 4-х в каждом, признак помещается в одну строку
for col in cat_cols:
  generate_pies(data, col)
```

Для one-hot-encoding'a можно испльзовать комбинацию LabelEncoder + OneHotEncoder (сначала превращаем строчки в числа, а потом числа првращаем в one-hot вектора.)

```python
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, RobustScaler, LabelEncoder, OneHotEncoder
from sklearn.pipeline import make_pipeline
from sklearn.compose import ColumnTransformer
```

```python
# Преобразование бинарных, категориальных признаков/нормировка числовых признаков
# Подготовим обучающий датасет
X = data.drop(target_col, axis=1)
y = data[target_col]
# С учётом, что разные преобразования к разным столбцам, то используем ColumnTransformer
# Для бинарных удаляем лишний столбец после OHE
col_transformer = ColumnTransformer(
    [('num', StandardScaler(), num_cols),
    ('cat', OneHotEncoder(drop='if_binary', sparse_output=False), cat_cols)],
    remainder='passthrough')

pipe = make_pipeline(col_transformer, LogisticRegression(random_state=42))
gd_search = GridSearchCV(pipe, param_grid={'logisticregression__C': [100, 10, 1, 0.1, 0.01, 0.001]}, cv=5, scoring='roc_auc', refit=True)
gd_search.fit(X, y)

# Попробуем с балансировкой классов
pipe_balanced = make_pipeline(col_transformer, LogisticRegression(random_state=42, class_weight = 'balanced'))
gd_search_balanced = GridSearchCV(pipe_balanced, param_grid={'logisticregression__C': [100, 10, 1, 0.1, 0.01, 0.001]}, cv=5, scoring='roc_auc', refit=True)
gd_search_balanced.fit(X, y)

print("Лучшее значение параметра C: ", gd_search.best_params_['logisticregression__C'])
print("Лучшее значение метрики roc_auc: ", gd_search.best_score_)

print("Лучшее значение параметра C (балансировка классов): ", gd_search_balanced.best_params_['logisticregression__C'])
print("Лучшее значение метрики roc_auc (балансировка классов): ", gd_search_balanced.best_score_)
```
